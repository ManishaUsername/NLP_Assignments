Assignment 08: Apply WordNet-based Word Sense Disambiguation to improve meaning interpretation in
ambiguous sentences

Step 1: Import Required Libraries

```python
import nltk
from nltk.corpus import wordnet as wn
from nltk.wsd import lesk
from nltk.tokenize import word_tokenize
```

STEP 2: Download Required NLTK Resources

---



```python
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

STEP 3: Define the Ambiguous Sentence



```python
sentence = "He went to the bank to deposit money"
```

STEP 4: Tokenize the Sentence

```python
tokens = word_tokenize(sentence)
print(tokens)
```

STEP 5: Select the Target (Ambiguous) Word

```python
target_word = "bank"
```

STEP 6: Apply the Lesk Algorithm

```python
sense = lesk(tokens, target_word)
```

STEP 7: Display the Selected Sense

```python
print("Best Sense:", sense)
print("Definition:", sense.definition())
```

STEP 8: Show All Possible Senses

```python
all_senses = wn.synsets(target_word)

for s in all_senses:
    print(s, ":", s.definition())
```
