Assignment 09: Use appropriate algorithm and perform sentiment analysis for Indian Language.

Indian Language: Hindi

Algorithm: Lexicon-Based Sentiment Analysis

STEP 1: Define Positive and Negative Word Lists

```python
positive_words = [
    "अच्छा", "अच्छी", "बेहतरीन", "शानदार",
    "खुश", "सुंदर", "पसंद", "उत्कृष्ट"
]

negative_words = [
    "खराब", "बुरा", "बेकार", "गंदा",
    "दुखी", "नफरत", "घटिया"
]
```

STEP 2: Take Input Sentence (Hindi)

```python
sentence = "यह फिल्म बहुत अच्छी है"
```

STEP 3: Tokenize the Sentence

```python
tokens = sentence.split()
print("Tokens:", tokens)
```

STEP 4: Initialize Sentiment Scores

```python
positive_score = 0
negative_score = 0
```

STEP 5: Compare Tokens with Sentiment Lexicon

```python
for word in tokens:
    if word in positive_words:
        positive_score += 1
    elif word in negative_words:
        negative_score += 1
```

STEP 6: Decide Final Sentiment

```python
if positive_score > negative_score:
    sentiment = "Positive"
elif negative_score > positive_score:
    sentiment = "Negative"
else:
    sentiment = "Neutral"
```

STEP 7: Display Output

```python
print("Sentence:", sentence)
print("Positive Score:", positive_score)
print("Negative Score:", negative_score)
print("Final Sentiment:", sentiment)
```
